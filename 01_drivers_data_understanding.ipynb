## 5. Placeholder tersembunyi

Nilai kosong yang ditulis sebagai teks (`na`, `unknown`, `-`, dan sejenisnya) tidak terhitung sebagai kosong oleh pandas. Kode berikut mencari kandidatnya di semua kolom teks.

```python
CANDIDATES = {"", "null", "n/a", "na", "-", "--", "unknown", "nan", "nil", "missing", "?"}
found = {}
for c in df.select_dtypes(exclude="number").columns:
    s = df[c].dropna().astype(str).str.strip().str.lower()
    hit = s[s.isin(CANDIDATES)].value_counts()
    if len(hit):
        found[c] = hit.to_dict()

pd.DataFrame(found).T.fillna(0).astype(int)
```

```python
# 'rating belum ada' ditulis sebagai teks "unrated". Apakah driver yang unrated punya trip lebih sedikit?
unrated = df["overall_driver_rating"].str.strip().str.lower().eq("unrated")
df.groupby(unrated)["total_trips_completed"].describe()
```

## 8. Kolom numerik yang tersimpan sebagai teks

`age`, `overall_driver_rating`, dan `avg_daily_earnings_INR` terbaca sebagai teks. Pola di bawah mengganti setiap angka dengan `9` supaya variasi format terlihat.

```python
for c in ["age", "overall_driver_rating", "avg_daily_earnings_INR"]:
    pola = df[c].dropna().astype(str).str.strip().str.replace(r"\d+", "9", regex=True)
    print(c)
    print(pola.value_counts().head(10), "\n")
```

```python
age_num = df["age"].str.extract(r"(-?\d+\.?\d*)")[0].astype(float)
valid = age_num[(age_num >= 0) & (age_num != 999)]

print("terisi:", df["age"].notna().sum(), "| angka berhasil diambil:", age_num.notna().sum())
print("negatif:", (age_num < 0).sum(), "| bernilai 999:", (age_num == 999).sum())
print("rentang umur valid:", valid.min(), "sampai", valid.max())
```

## 10. Tanggal

Kolom `interview_date` dan `onboarding_date` memakai beberapa format sekaligus. Pola di bawah mengganti angka dengan `9` dan huruf dengan `A` untuk menghitung ragam formatnya.

```python
def bentuk(s):
    s = s.dropna().astype(str).str.strip()
    return s.str.replace(r"\d", "9", regex=True).str.replace(r"[A-Za-z]+", "A", regex=True)

for c in ["interview_date", "onboarding_date"]:
    print(c)
    print(bentuk(df[c]).value_counts().to_string(), "\n")
```

Untuk format angka seperti `03/05/22`, urutan hari dan bulan harus dipastikan dari rentang nilainya.

```python
s = df["onboarding_date"].dropna().astype(str).str.strip()
shape = bentuk(s)
num_shapes = ["99/99/99", "99-99-9999", "99.99.9999", "9999/99/99"]
num = s[shape.isin(num_shapes)]
parts = num.str.extract(r"^(\d+)[/.\-](\d+)[/.\-](\d+)$").astype(int)
parts.columns = ["p1", "p2", "p3"]
parts["shape"] = shape[num.index]
parts.groupby("shape")[["p1", "p2", "p3"]].agg(["min", "max"])
```

Pada `99/99/99`, bagian pertama maksimal 12, jadi urutannya bulan/hari/tahun. Pada `99-99-9999` dan `99.99.9999`, bagian pertama sampai 31, jadi urutannya hari-bulan-tahun.
